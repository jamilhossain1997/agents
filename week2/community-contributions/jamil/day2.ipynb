```python
load_dotenv(override=True)
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
GML_API_KEY = os.getenv("GML_API_KEY")

if GEMINI_API_KEY:
    print(f"Gemini API Key exists and begins with:{GEMINI_API_KEY[:8]}")
else:
    print("Gemini API Key does not exist.")
    
if OPENROUTER_API_KEY:
    print(f"OpenRouter API Key exists and begins with:{OPENROUTER_API_KEY[:8]}")
else:
    print("OpenRouter API Key does not exist.")

if GROQ_API_KEY:
    print(f"GROQ API Key exists and begins with:{GROQ_API_KEY[:8]}")
else:
    print("GROQ API Key does not exist.")
    
if GML_API_KEY:
    print(f"GML API Key exists and begins with:{GML_API_KEY[:8]}")
else:
    print("GML API Key does not exist.") 
```

```python
ollama_url = "http://localhost:11434/v1"
openrouter_url = "https://openrouter.ai/api/v1"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
groq_url = "https://api.groq.com/openai/v1"
gml_url="https://api.z.ai/api/paas/v4/"
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
openrouter = OpenAI(api_key=OPENROUTER_API_KEY, base_url=openrouter_url)
gemini = OpenAI(api_key=GEMINI_API_KEY, base_url=gemini_url)
groq = OpenAI(api_key=GROQ_API_KEY, base_url=groq_url)
gml = OpenAI(api_key=GML_API_KEY, base_url=gml_url)
```

```python
system_message = "You are a helpful assistant"

def generate_response(client, model_name, prompt):
    messages =[
        {"role": "system", "content": system_message},
        {"role":"user" , "content" : prompt}
    ]
    
    response = client.chat.completions.create(
        model=model_name,
        messages=messages,
        max_tokens=200
    )
    
    return response.choices[0].message.content
    
    
```

```python
generate_response(openrouter, "openai/gpt-4o-mini", "What is today's date?")
```

```python

from datetime import datetime

today = datetime.now().astimezone().strftime("%B %d, %Y")

prompt = f"""
Today's date is {today}.

Question: What is today's date?
Answer using the date provided above.
"""

response = generate_response(
    ollama,
    "llama3.2:latest",
    prompt
)

print(response)
```
