```python
system_message = "You are a helpful assistant"

def generate_response(client, model_name, prompt):
    messages =[
        {"role": "system", "content": system_message},
        {"role":"user" , "content" : prompt}
    ]
    
    response = client.chat.completions.create(
        model=model_name,
        messages=messages,
        max_tokens=200
    )
    
    return response.choices[0].message.content
    
    
```

```python
generate_response(openrouter, "openai/gpt-4o-mini", "What is today's date?")
```

```python
system_message = "You are a helpful assistant that responds in markdown without code blocks"
def llama3_message(prompt):
    messages = [
        {"role": "system" , "content": system_message},
        {"role": "user" , "content" : prompt}
    ]
    
    response = ollama.chat.completions.create(
        messages=messages,
        model="llama3.2:latest",
    )
    
    return response.choices[0].message.content
```

```python
in_message = gr.Textbox(label="your message" , info="Enter a message for LLama3.2" , lines=7)
out_message = gr.Markdown(label="Response:")

view = gr.Interface(
    fn=llama3_message,
    inputs=in_message,
    outputs=out_message,
    title="llama3_message",
    examples=[
        "Explain the Transformer architecture to a layperson",
        "Explain the Transformer architecture to an aspiring AI engineer",
    ], 
    flagging_mode="never"

)

view.launch(share=True)
```

```python
# Let's create a call that streams back results
# If you'd like a refresher on Generators (the "yield" keyword),
# Please take a look at the Intermediate Python guide in the guides folder

def stream_llama(prompt):
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
      ]
    stream = ollama.chat.completions.create(
        model='llama3.2:latest',
        messages=messages,
        stream=True
    )
    result = ""
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        yield result
```

```python

import gradio as gr

# Your existing function: llama3_message()

custom_css = """
.gradio-container {
    max-width: 1000px !important;
    margin: auto !important;
    font-family: 'Inter', sans-serif;
}

#app-header {
    text-align: center;
    padding: 25px 15px 15px;
}

#app-header h1 {
    font-size: 34px;
    font-weight: 750;
    margin-bottom: 8px;
}

#app-header p {
    font-size: 15px;
    opacity: 0.8;
}

#chat-input textarea {
    font-size: 16px !important;
    border-radius: 14px !important;
}

#response-box {
    border-radius: 14px !important;
    min-height: 250px;
}

#generate-btn {
    border-radius: 12px !important;
    font-weight: 600 !important;
}

footer {
    display: none !important;
}
"""

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="violet",
        secondary_hue="blue",
        neutral_hue="slate",
    ),
    css=custom_css,
    title="Llama 3.2 AI Assistant",
) as view:

    gr.HTML("""
        <div id="app-header">
            <h1>🦙 Llama 3.2 AI Assistant</h1>
            <p>Your personal AI assistant, powered by Llama 3.2</p>
        </div>
    """)

    with gr.Row():
        with gr.Column(scale=1):
            in_message = gr.Textbox(
                label="Your Message",
                placeholder="Ask anything... e.g. Explain how transformers work.",
                info="Enter a question for Llama 3.2",
                lines=7,
                max_lines=12,
                elem_id="chat-input",
            )

            gr.Examples(
                examples=[
                    ["Explain the Transformer architecture to a layperson."],
                    ["Explain Transformers to an aspiring AI engineer."],
                    ["What is Retrieval-Augmented Generation (RAG)?"],
                    ["Write a Python function to reverse a string."],
                ],
                inputs=in_message,
                label="✨ Try an example",
            )

            with gr.Row():
                submit_btn = gr.Button(
                    "✨ Generate Response",
                    variant="primary",
                    elem_id="generate-btn",
                )
                clear_btn = gr.ClearButton(
                    components=[in_message],
                    value="Clear Input",
                )

        with gr.Column(scale=1):
            out_message = gr.Markdown(
                value="### 🤖 AI Response\n\nYour response will appear here...",
                label="Response",
                elem_id="response-box",
            )

    submit_btn.click(
        fn=stream_llama,
        inputs=in_message,
        outputs=out_message,
        show_progress="full",
    )

    in_message.submit(
        fn=stream_llama,
        inputs=in_message,
        outputs=out_message,
        show_progress="full",
    )

    gr.HTML("""
        <div style="text-align:center; padding:20px; opacity:0.65;">
            Built with ❤️ using Gradio and Llama 3.2
        </div>
    """)

if __name__ == "__main__":
    view.launch(share=True, inbrowser=True)
```
