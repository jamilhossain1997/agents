```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
GROQ_API_KEY = os.getenv("GROQ_API_KEY")

for name, key in [("OpenAI", OPENAI_API_KEY),
                  ("OpenRouter", OPENROUTER_API_KEY),
                  ("Gemini", GEMINI_API_KEY),
                  ("Groq", GROQ_API_KEY)]:
    print(f"{name} API key is {'set' if key else 'NOT set'}")
```

```python
openrouter_url = "https://openrouter.ai/api/v1"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
groq_url = "https://api.groq.com/openai/v1"
ollama_url = "http://localhost:11434/v1"



clients = {
    "openrouter": OpenAI(api_key=OPENROUTER_API_KEY, base_url=openrouter_url),
    "gemini": OpenAI(api_key=GEMINI_API_KEY, base_url=gemini_url),
    "groq": OpenAI(api_key=GROQ_API_KEY, base_url=groq_url),
    "ollama": OpenAI(api_key="ollama", base_url=ollama_url),
}
```

```python
tell_a_joke = [
    {"role": "user", "content": "Tell a joke for a student on the journey to becoming an expert in LLM Engineering"},
]
```

```python
responses = clients["groq"].chat.completions.create(
    model= "openai/gpt-oss-20b",
    messages= tell_a_joke
)

print(responses.choices[0].message.content)
```

```python
responses = clients["gemini"].chat.completions.create(
    model="gemini-3.5-flash",
    messages=tell_a_joke
)

display(Markdown(f"**Gemini Response:** {responses.choices[0].message.content}"))
```
