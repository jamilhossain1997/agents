```python
import os
from dotenv import load_dotenv;

load_dotenv(override=True)

api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    print("Error: GEMINI_API_KEY is not set in this Enviroment.Please set it in your .env file or as an enviroment variable.")
elif not api_key.strip() != "api_key":
        print("Error: GEMINI_API_KEY is not set in this Enviroment.Please set it in your .env file or as an enviroment variable.")
else:
      print("GEMINI_API_KEY is set correctly.")
```

```python
import requests;

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

payload = {
    "model": "gemini-3.5-flash",
    "messages": [
        {"role": "user", "content": "Tell me a fun fact"}]
}

payload
```

```python
response = requests.post(
    "https://generativelanguage.googleapis.com/v1beta/openai/chat/completions",
    headers=headers,
    json=payload
)

print(response.json())
print(response.status_code)
print(response.json()["choices"][0]["message"]["content"])
```

what is package use gemini and chatGpt ?

It's knows paython client libary
It's nothing more than a wrapper around making this exact call to the http endpoint.
It just allows you to work with nice Python code instead of messing around with janky json objects.
But that's it. It's open-source and lightweight. Some people think it contains OpenAI and others  model code - it doesn't!

```python
from google import genai

client = genai.Client(api_key=api_key)

messages = [
    {"role": "user", "content": "Tell me a fun fact"}
]

# convert OpenAI-style messages -> genai contents
contents = [
    {"role": m["role"], "parts": [{"text": m["content"]}]}
    for m in messages
]

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=contents
)

print(response.text)
```

```python
from openai import OpenAI

base_url="https://generativelanguage.googleapis.com/v1beta/openai/"

openai_client = OpenAI(api_key=api_key, base_url=base_url)

response =openai_client.chat.completions.create(
    model="gemini-3.5-flash",
    messages=[
        {"role": "user", "content": "Tell me a fun fact"}
    ]
)

print(response.choices[0].message.content)
```
